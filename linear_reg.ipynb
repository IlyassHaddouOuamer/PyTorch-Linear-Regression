# Linear Regression 1D prediction

```python
import torch
```

### Prediction

```python
# Define w=2 and b=-1 for y = wx + b
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(-1.0, requires_grad=True)
```

```python
# Function forward(x) for prediction
def forward(x):
    y = w * x + b
    return y
```

```python
# Predict the output for x=1
x = torch.tensor([[1.0]])
yhat = forward(x)
print(yhat)  
```

```python
# Create x tensor and check the shape of x tensor
x = torch.tensor([[1.0], [2.0], [3.0]])
print(x.shape)
```

```python
# make the prediction
yhat = forward(x)
print(yhat)
```

# Class Linear

```python
from torch.nn import Linear
torch.manual_seed(1)
```

```python
# Create Linear Regression model
model = Linear(in_features=1, out_features=1,bias=True)
print("Parameters w and b :", list(model.parameters()))
```

```python
print("python dictionary:", model.state_dict())
print("keys:", model.state_dict().keys())
print("values:", model.state_dict().values())
```

```python
print("weight:",model.weight)
print("bias:",model.bias)
```

```python
# make the prediction
x = torch.tensor([[1.0]])
yhat = model(x)
print(yhat)
```

```python
# Create the prediction
x = torch.tensor([[1.0], [2.0], [3.0]])
yhat = model(x)
print(yhat)
```

### Build custom modeles

```python
from torch import nn
```

```python
class LR(nn.Module):
    
    # Constructor
    def __init__(self, input_dim, output_dim):
        super(LR, self).__init__()
        # define the linear layer
        self.linear = nn.Linear(input_dim, output_dim)

    def forward(self, x):
        return self.linear(x)
```

```python
# Create the linear regression model
lr_model = LR(input_dim=1, output_dim=1)
print("Parameters w and b :", list(lr_model.parameters()))
print("Linear model :", lr_model)
```

```python
# Try our customized model
x = torch.tensor([[1.0]])
yhat = lr_model(x)
print(yhat)
```

```python
x = torch.tensor([[1.0], [2.0], [3.0]])
yhat = lr_model(x)  
print(yhat)
```

```python
print("python dictionary:", lr_model.state_dict())
print("keys:", lr_model.state_dict().keys())
print("values:", lr_model.state_dict().values())
```
